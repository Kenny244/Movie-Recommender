# Saved Notebook Outputs

This notebook gathers and displays every output saved in `01_data_exploration.ipynb`. It reads the notebook file without executing its cells, so model training is not repeated. Run the Python cell below to refresh the displayed results after saving outputs in the analysis notebook.

```python
from pathlib import Path
import json
from IPython.display import display

source_path = Path("01_data_exploration.ipynb")
if not source_path.exists():
    source_path = Path.cwd() / "notebooks" / "01_data_exploration.ipynb"

if not source_path.exists():
    raise FileNotFoundError(
        "Could not find notebooks/01_data_exploration.ipynb. "
        "Open this notebook from the project workspace and try again."
    )

with source_path.open(encoding="utf-8") as notebook_file:
    source_notebook = json.load(notebook_file)

saved_output_count = 0
for cell_number, cell in enumerate(source_notebook.get("cells", []), start=1):
    outputs = cell.get("outputs", [])
    if not outputs:
        continue

    source = cell.get("source", [])
    source_text = "".join(source) if isinstance(source, list) else source
    source_summary = " ".join(source_text.strip().splitlines())
    print(f"Cell {cell_number}: {source_summary[:180] or '[no source text]'}")

    for output in outputs:
        output_type = output.get("output_type")
        if output_type == "stream":
            stream_text = output.get("text", "")
            print("".join(stream_text) if isinstance(stream_text, list) else stream_text)
        elif output_type in {"display_data", "execute_result"}:
            data = output.get("data", {})
            if data:
                display(data, raw=True)
        elif output_type == "error":
            print(f"{output.get('ename', 'Error')}: {output.get('evalue', '')}")
            for traceback_line in output.get("traceback", []):
                print(traceback_line)
        saved_output_count += 1

if saved_output_count == 0:
    print(
        "No saved outputs were found. Run the analysis notebook cells, "
        "save the notebook, then rerun this cell."
    )
else:
    print(f"\nDisplayed {saved_output_count} saved output(s).")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from bs4 import BeautifulSoup
from IPython.display import display

comparison_cell = None
for cell in source_notebook.get("cells", []):
    cell_source = cell.get("source", [])
    cell_source = "".join(cell_source) if isinstance(cell_source, list) else cell_source
    if "comparison_with_rl = pd.concat" in cell_source:
        comparison_cell = cell
        break

if comparison_cell is None:
    raise ValueError("The saved RL comparison cell was not found in the analysis notebook.")

comparison_html = next(
    (
        output["data"]["text/html"]
        for output in comparison_cell.get("outputs", [])
        if "text/html" in output.get("data", {})
    ),
    None,
)
if comparison_html is None:
    raise ValueError(
        "The comparison has no saved HTML output. Run and save the final comparison cell first."
    )

soup = BeautifulSoup("".join(comparison_html), "html.parser")
table_rows = [
    [cell.get_text(strip=True) for cell in row.find_all(["th", "td"])]
    for row in soup.find_all("tr")
]
headers = table_rows[0]
records = [dict(zip(headers, row)) for row in table_rows[1:] if len(row) == len(headers)]
comparison_data = pd.DataFrame(records)

models_to_compare = ["Popularity", "GRU + offline CQL"]
metric_columns = ["Precision@10", "Recall@10", "NDCG@10"]
plot_data = comparison_data.set_index("Model").loc[models_to_compare, metric_columns]
plot_data = plot_data.astype(float).rename(
    index={
        "Popularity": "Traditional: Popularity",
        "GRU + offline CQL": "Deep RL: GRU + CQL",
    }
)

fig, ax = plt.subplots(figsize=(7.2, 4.5), layout="constrained")
positions = range(len(metric_columns))
bar_width = 0.36
traditional_values = plot_data.iloc[0].to_numpy()
rl_values = plot_data.iloc[1].to_numpy()

traditional_bars = ax.bar(
    [position - bar_width / 2 for position in positions],
    traditional_values,
    bar_width,
    label=plot_data.index[0],
    color="#286B8E",
)
rl_bars = ax.bar(
    [position + bar_width / 2 for position in positions],
    rl_values,
    bar_width,
    label=plot_data.index[1],
    color="#D77A3D",
)

ax.bar_label(traditional_bars, fmt="%.3f", padding=3, fontsize=9)
ax.bar_label(rl_bars, fmt="%.3f", padding=3, fontsize=9)
ax.set_xticks(list(positions), metric_columns)
ax.set_ylabel("Score (higher is better)")
ax.set_ylim(0, max(traditional_values.max(), rl_values.max()) * 1.22)
ax.set_title(
    "Popularity Baseline vs. Sequential Deep RL\n"
    "MovieLens 1M chronological holdout; exposure logs unavailable",
    loc="left",
    weight="bold",
    fontsize=12,
    pad=12,
)
ax.legend(frameon=False, ncols=2, loc="upper right")
ax.set_axisbelow(True)
ax.yaxis.grid(True, linestyle="--", linewidth=0.7, alpha=0.35)
ax.spines[["top", "right"]].set_visible(False)

project_root = source_path.resolve().parent.parent
results_dir = project_root / "results"
results_dir.mkdir(exist_ok=True)
png_path = results_dir / "popularity_vs_gru_cql.png"
pdf_path = results_dir / "popularity_vs_gru_cql.pdf"
fig.savefig(png_path, dpi=300, bbox_inches="tight")
fig.savefig(pdf_path, bbox_inches="tight")
display(fig)
print(f"Saved PNG: {png_path}")
print(f"Saved PDF: {pdf_path}")
```
